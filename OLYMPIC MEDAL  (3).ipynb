```python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import classification_report, accuracy_score
```

```python
enc=LabelEncoder()
```

```python
data["City"]=enc.fit_transform(data["City"])
data["Discipline"]=enc.fit_transform(data["Discipline"])
data["Event"]=enc.fit_transform(data["Event"])
data["Athlete"]=enc.fit_transform(data["Athlete"])
data["Gender"]=enc.fit_transform(data["Gender"])
data["Country_Code"]=enc.fit_transform(data["Country_Code"])
data["Country"]=enc.fit_transform(data["Country"])
data["Event_gender"]=enc.fit_transform(data["Event_gender"])
```

```python
data["Sport"]=enc.fit_transform(data["Sport"])
```

```python
from sklearn.model_selection import train_test_split
```

```python
x_train,x_test,y_train,y_test =train_test_split(x,y,test_size=0.2,random_state=10)
```

```python
x_train.shape
```

```python
x_test.shape
```

```python
from sklearn.tree import DecisionTreeClassifier,plot_tree
```

```python
model=DecisionTreeClassifier()
```

```python
model.fit(x_train,y_train)
```

```python
model.score(x_train,y_train)
```

```python
model.score(x_test,y_test)
```

```python
plot_tree(model,filled=True,feature_names=x.columns, class_names=["0","1"],max_depth=1)
```

```python
# rendom forest tree model
```

```python
from sklearn.ensemble import RandomForestClassifier 
```

```python
model=RandomForestClassifier(random_state=20)
```

```python
#prediction 
```

```python
y_pred = model.predict(x_test)
```

```python
print("Accuracy:",accuracy_score(y_test, y_pred))
print(classification_report(y_test,y_pred ))
```
